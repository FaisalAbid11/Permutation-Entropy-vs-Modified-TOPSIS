```python
import pandas as pd
import scipy.stats as stats

# Load dataset
file_path = '/content/Final Primary_dataset_from_Indonesian_Private_University.csv'
df = pd.read_csv(file_path)

# Clean column names (strip spaces)
df.columns = df.columns.str.strip()

# Features and target
features = [
    'age', 'Gender', 'Work tenure', 'Education', 'Job position', 'working hour',
    'satisfaction with workload', 'satisfied  compensation ',
    'good relationship with peers', 'satisfied with career and  opportunity',
    'satisfied with job profession ', 'monthly salary',
    'satisfied with  work-life balance', 'work is meaningful',
    'family support', 'mentally well and do not have anxiety'
]
target = 'TOI (turnover intention)'

# Normalize feature names
features = [f.strip() for f in features]
target = target.strip()

# Chi-Square calculation
chi_results = []
for col in features:
    if col not in df.columns:
        continue
    contingency = pd.crosstab(df[col], df[target])
    chi2, p, _, _ = stats.chi2_contingency(contingency)
    chi_results.append([col, chi2, p])

# Create results table
chi_square_table = pd.DataFrame(
    chi_results, columns=["Feature", "Chi2", "p-value"]
).sort_values(by="Chi2", ascending=False).reset_index(drop=True)

# Show results
print(chi_square_table)
```

```python
# --- Setup ---
import pandas as pd
import numpy as np
from sklearn.feature_selection import f_classif
from sklearn.utils.validation import check_is_fitted

# 1) Load your CSV (adjust path if needed)
file_path = '/content/Final Primary_dataset_from_Indonesian_Private_University.csv'
df = pd.read_csv(file_path)

# 2) Clean column names a bit
df.columns = df.columns.str.strip()

# 3) Identify target and drop obvious junk columns if present
TARGET = 'TOI (turnover intention)'
drop_cols = ['Unnamed: 5']  # harmless empty column in your file
df = df.drop(columns=[c for c in drop_cols if c in df.columns], errors='ignore')

# 4) Guard: ensure target is 0/1 numeric
if not np.issubdtype(df[TARGET].dtype, np.number):
    # try to coerce
    df[TARGET] = pd.to_numeric(df[TARGET], errors='coerce')

# Drop rows where target is missing
df = df.dropna(subset=[TARGET])

# 5) Split X/y
X_raw = df.drop(columns=[TARGET])
y = df[TARGET].astype(int)

# 6) Decide which columns are numeric vs categorical.
#    - Numeric (Likert 1–5) stays numeric (good for ANOVA F)
#    - Non-numeric becomes one-hot encoded
numeric_cols = X_raw.select_dtypes(include=[np.number]).columns.tolist()
categorical_cols = [c for c in X_raw.columns if c not in numeric_cols]

# 7) One-hot encode categoricals with a SAFE separator so we can group back later
SAFE_SEP = '§'  # unlikely to appear in names
X_cat = pd.get_dummies(
    X_raw[categorical_cols],
    prefix=categorical_cols,
    prefix_sep=SAFE_SEP,
    drop_first=True,   # drop a baseline level to avoid perfect collinearity
    dtype=np.uint8
) if categorical_cols else pd.DataFrame(index=X_raw.index)

# 8) Combine numeric + encoded categoricals
X_num = X_raw[numeric_cols].copy()
X = pd.concat([X_num, X_cat], axis=1)

# Drop any rows with NaNs in features (rare, but safest for f_classif)
mask = X.notna().all(axis=1)
X = X.loc[mask]
y = y.loc[mask]

# 9) Run ANOVA F (univariate)
f_values, p_values = f_classif(X.values, y.values)

per_column = pd.DataFrame({
    'Column': X.columns,
    'F_value': f_values,
    'p_value': p_values
}).sort_values('F_value', ascending=False).reset_index(drop=True)

# 10) Aggregate back to original "Feature" (for dummies),
#     keeping numeric columns as-is.
#     We detect original feature for dummies by splitting on SAFE_SEP.
def original_feature(col: str) -> str:
    # if it contains the SAFE_SEP, the part before it is the original feature
    return col.split(SAFE_SEP, 1)[0] if SAFE_SEP in col else col

per_column['Feature'] = per_column['Column'].map(original_feature)

# Aggregate strategy:
# - For each original feature with multiple dummies:
#   * take max F_value (strongest signal among categories)
#   * take min p_value (most significant among categories)
# - For numeric features (no dummies), they remain single rows.
agg = (per_column
       .groupby('Feature', as_index=False)
       .agg(F_value_max=('F_value', 'max'),
            p_value_min=('p_value', 'min'),
            n_columns=('Column', 'count'))
       .sort_values('F_value_max', ascending=False)
       .reset_index(drop=True))

# 11) Display top results
print("\n=== ANOVA F (per column / dummy) — top 20 ===")
print(per_column.head(20))

print("\n=== ANOVA F (grouped back to original feature) ===")
print(agg)


```

```python
import pandas as pd
import numpy as np
from sklearn.feature_selection import mutual_info_classif
from sklearn.preprocessing import OrdinalEncoder

# === 1) Load & normalize column names ===
file_path = '/content/Final Primary_dataset_from_Indonesian_Private_University.csv'  # <- adjust if needed
df = pd.read_csv(file_path)
df.columns = (
    df.columns
      .str.strip()
      .str.replace(r'\s+', ' ', regex=True)
)

# === 2) Your features & target (normalized names) ===
features = [
    'age', 'Gender', 'Work tenure', 'Education', 'Job position', 'working hour',
    'satisfaction with workload', 'satisfied compensation',
    'good relationship with peers', 'satisfied with career and opportunity',
    'satisfied with job profession', 'monthly salary',
    'satisfied with work-life balance', 'work is meaningful',
    'family support', 'mentally well and do not have anxiety'
]
target = 'TOI (turnover intention)'

# Keep only requested columns and drop rows missing target
df = df[features + [target]].dropna(subset=[target])

# === 3) Split categorical vs numeric from YOUR list ===
categorical_features = [
    'age', 'Gender', 'Work tenure', 'Education', 'Job position',
    'working hour', 'monthly salary'
]
numeric_features = [c for c in features if c not in categorical_features]

# Prepare X: encode categoricals, keep numeric as numeric
X_cat = df[categorical_features].astype('category')
X_num = df[numeric_features].apply(pd.to_numeric, errors='coerce')

# Drop rows with NaNs in numeric features (keeps MI happy)
row_mask = X_num.notna().all(axis=1)
X_cat = X_cat.loc[row_mask]
X_num = X_num.loc[row_mask]
y = df.loc[row_mask, target].astype(int)

enc = OrdinalEncoder(handle_unknown='use_encoded_value', unknown_value=-1)
X_cat_enc = pd.DataFrame(
    enc.fit_transform(X_cat),
    columns=categorical_features,
    index=X_cat.index
).astype(int)

# Combine back
X = pd.concat([X_cat_enc, X_num], axis=1)

# Build the discrete_features mask (True for categoricals, False for numeric)
discrete_mask = [col in categorical_features for col in X.columns]

# === 4) Information Gain (Mutual Information) ===
IG_scores = mutual_info_classif(
    X.values, y.values,
    discrete_features=discrete_mask,
    random_state=42
)

ig_table = pd.DataFrame({
    "Feature": X.columns,
    "Information_Gain": IG_scores,
    "Type": ["categorical" if d else "numeric" for d in discrete_mask]
}).sort_values("Information_Gain", ascending=False).reset_index(drop=True)

print(ig_table)

# Optional: save to CSV
ig_table.to_csv('/content/information_gain_scores.csv', index=False)
print("\nSaved: /content/information_gain_scores.csv")
```

```python
# ===== LASSO (L1 Logistic) feature selection for TOI =====
# - Uses ONLY your declared features + target
# - One-hot encodes categoricals, scales Likert numerics
# - Cross-validated strength of regularization (C)
# - Produces per-column and grouped-by-feature tables and saves as CSVs

import pandas as pd
import numpy as np
import sklearn
from sklearn.model_selection import StratifiedKFold
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegressionCV

print("scikit-learn version:", sklearn.__version__)

# ---------- 1) Load & normalize column names ----------
file_path = '/content/Final Primary_dataset_from_Indonesian_Private_University.csv'  # <-- adjust if needed
df = pd.read_csv(file_path)
# strip ends & collapse multiple spaces to a single space (your CSV has extra spaces)
df.columns = df.columns.str.strip().str.replace(r'\s+', ' ', regex=True)

# ---------- 2) Your features and target (normalized names) ----------
features = [
    'age', 'Gender', 'Work tenure', 'Education', 'Job position', 'working hour',
    'satisfaction with workload', 'satisfied compensation',
    'good relationship with peers', 'satisfied with career and opportunity',
    'satisfied with job profession', 'monthly salary',
    'satisfied with work-life balance', 'work is meaningful',
    'family support', 'mentally well and do not have anxiety'
]
target = 'TOI (turnover intention)'

# keep only your columns; drop rows with missing target
missing = [c for c in features + [target] if c not in df.columns]
if missing:
    raise KeyError(f"Missing columns after normalization: {missing}")

df = df[features + [target]].dropna(subset=[target])
y = df[target].astype(int)

# ---------- 3) Split numeric (Likert) vs categorical ----------
categorical_features = [
    'age', 'Gender', 'Work tenure', 'Education', 'Job position',
    'working hour', 'monthly salary'
]
numeric_features = [c for c in features if c not in categorical_features]

# ensure Likert numerics are numeric
for col in numeric_features:
    df[col] = pd.to_numeric(df[col], errors='coerce')

# drop rows with any missing in selected features
df = df.dropna(subset=features)
X = df[features]
y = df[target].astype(int)

# ---------- 4) Preprocessing: OHE for categoricals + scaling for numerics ----------
SAFE_SEP = '§'  # used to rebuild original feature names from dummies later

# Handle OneHotEncoder API differences across sklearn versions
# - <1.2: OneHotEncoder(..., sparse=False)
# - >=1.2: OneHotEncoder(..., sparse_output=False)
ohe_kwargs = dict(drop='first', handle_unknown='ignore')
ver = tuple(int(v) for v in sklearn.__version__.split('.')[:2])
if ver >= (1, 2):
    ohe_kwargs['sparse_output'] = False
else:
    ohe_kwargs['sparse'] = False

ohe = OneHotEncoder(**ohe_kwargs)
scaler = StandardScaler()

preprocess = ColumnTransformer(
    transformers=[
        ('cat', ohe, categorical_features),
        ('num', scaler, numeric_features),
    ],
    remainder='drop'
)

# ---------- 5) L1-penalized Logistic Regression with CV ----------
# We optimize F1 and use class_weight='balanced' for any class imbalance.
logreg_cv = LogisticRegressionCV(
    Cs=np.logspace(-2, 1, 8),  # search range for C (inverse of regularization)
    cv=StratifiedKFold(n_splits=3, shuffle=True, random_state=42),
    penalty='l1',
    solver='saga',             # supports l1 with class_weight
    scoring='f1',
    class_weight='balanced',
    max_iter=4000,
    n_jobs=-1,
    refit=True
)

pipe = Pipeline(steps=[('prep', preprocess), ('clf', logreg_cv)])
pipe.fit(X, y)

# ---------- 6) Recover feature names after preprocessing ----------
# Build Feature§Category names for one-hot columns so we can group later.
cat_feature_names = []
ohe_fitted = pipe.named_steps['prep'].named_transformers_['cat']
for feat, cats in zip(categorical_features, ohe_fitted.categories_):
    # drop='first' => first category is the baseline and not present in output
    for cat in cats[1:]:
        cat_feature_names.append(f"{feat}{SAFE_SEP}{cat}")

num_feature_names = numeric_features[:]  # unchanged by StandardScaler
all_feature_names = cat_feature_names + num_feature_names

# ---------- 7) Coefficients (LASSO shrinkage; zeros mean dropped) ----------
coefs = pipe.named_steps['clf'].coef_.ravel()
per_column = pd.DataFrame({
    'Column': all_feature_names,
    'Coefficient': coefs
})
per_column['Abs_Coeff'] = per_column['Coefficient'].abs()
per_column = per_column.sort_values('Abs_Coeff', ascending=False).reset_index(drop=True)

def original_feature_name(col: str) -> str:
    return col.split(SAFE_SEP, 1)[0] if SAFE_SEP in col else col

per_column['Feature'] = per_column['Column'].map(original_feature_name)

# Aggregate back to original feature:
# - Importance = sum of |coefficients| over its columns
# - NonZero_Columns = count of nonzero coefficients (survived lasso)
# - Max_Abs = largest |coef| among its columns
# - Columns = total columns contributed (1 for numeric; >1 for categoricals)
grouped = (
    per_column
    .groupby('Feature', as_index=False)
    .agg(Importance=('Abs_Coeff', 'sum'),
         NonZero_Columns=('Coefficient', lambda x: int(np.sum(x != 0))),
         Max_Abs=('Abs_Coeff', 'max'),
         Columns=('Column', 'count'))
    .sort_values('Importance', ascending=False)
    .reset_index(drop=True)
)

# ---------- 8) Show & save ----------
print("\n=== L1-Logistic (LASSO) — Per Column Coefficients (top 25) ===")
print(per_column.head(25))
print("\n=== L1-Logistic (LASSO) — Grouped by Feature ===")
print(grouped)

per_column.to_csv('/content/lasso_per_column.csv', index=False)
grouped.to_csv('/content/lasso_grouped_by_feature.csv', index=False)
print("\nSaved CSVs:\n  /content/lasso_per_column.csv\n  /content/lasso_grouped_by_feature.csv")

# Optional: quick glance at class balance
print("\nClass counts (TOI):")
print(y.value_counts())
```

```python
import pandas as pd
import numpy as np
import sklearn
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.ensemble import RandomForestClassifier
from sklearn.inspection import permutation_importance

# 1) Load & clean
df = pd.read_csv('/content/Final Primary_dataset_from_Indonesian_Private_University.csv')
df.columns = df.columns.str.strip().str.replace(r'\s+', ' ', regex=True)

features = [
    'age', 'Gender', 'Work tenure', 'Education', 'Job position', 'working hour',
    'satisfaction with workload', 'satisfied compensation',
    'good relationship with peers', 'satisfied with career and opportunity',
    'satisfied with job profession', 'monthly salary',
    'satisfied with work-life balance', 'work is meaningful',
    'family support', 'mentally well and do not have anxiety'
]
target = 'TOI (turnover intention)'

df = df[features + [target]].dropna(subset=[target])
y = df[target].astype(int)

# 2) Split categorical vs numeric (Likert)
categorical = ['age','Gender','Work tenure','Education','Job position','working hour','monthly salary']
numeric = [c for c in features if c not in categorical]
for c in numeric:
    df[c] = pd.to_numeric(df[c], errors='coerce')
df = df.dropna(subset=features)

X = df[features]; y = df[target].astype(int)

# 3) Train/validation split (permutation importance on val set)
X_tr, X_val, y_tr, y_val = train_test_split(X, y, test_size=0.3, stratify=y, random_state=42)

# 4) One-hot encode categoricals (dense)
ohe_kwargs = dict(drop='first', handle_unknown='ignore')
ver = tuple(int(v) for v in sklearn.__version__.split('.')[:2])
if ver >= (1, 2):
    ohe_kwargs['sparse_output'] = False
else:
    ohe_kwargs['sparse'] = False

ohe = OneHotEncoder(**ohe_kwargs)
pre = ColumnTransformer(
    transformers=[
        ('cat', ohe, categorical),
        ('num', 'passthrough', numeric),
    ],
    remainder='drop'
)

# Fit transform
Xtr_enc = pre.fit_transform(X_tr)
Xval_enc = pre.transform(X_val)

# Encoded feature names (Feature§Category for dummies)
SAFE_SEP = '§'
cat_names = []
for feat, cats in zip(categorical, pre.named_transformers_['cat'].categories_):
    for cat in cats[1:]:  # drop='first' baseline omitted
        cat_names.append(f"{feat}{SAFE_SEP}{cat}")
enc_feature_names = cat_names + numeric

# 5) Random Forest
rf = RandomForestClassifier(
    n_estimators=400,
    max_features='sqrt',
    class_weight='balanced',
    random_state=42,
    n_jobs=-1
)
rf.fit(Xtr_enc, y_tr)

# 6) MDI importance
mdi = pd.DataFrame({
    'Column': enc_feature_names,
    'MDI_Importance': rf.feature_importances_
}).sort_values('MDI_Importance', ascending=False).reset_index(drop=True)

# 7) Permutation importance (on validation set) with F1
perm = permutation_importance(rf, Xval_enc, y_val, scoring='f1', n_repeats=20, random_state=42, n_jobs=-1)
perm_df = pd.DataFrame({
    'Column': enc_feature_names,
    'Perm_Importance': perm.importances_mean,
    'Perm_Std': perm.importances_std
}).sort_values('Perm_Importance', ascending=False).reset_index(drop=True)

# 8) Group back to original features
def orig(col): return col.split(SAFE_SEP, 1)[0] if SAFE_SEP in col else col
mdi['Feature'] = mdi['Column'].map(orig)
perm_df['Feature'] = perm_df['Column'].map(orig)

mdi_grouped = (mdi.groupby('Feature', as_index=False)
                 .agg(MDI_Importance_sum=('MDI_Importance','sum'),
                      Columns=('Column','count'))
                 .sort_values('MDI_Importance_sum', ascending=False))

perm_grouped = (perm_df.groupby('Feature', as_index=False)
                  .agg(Perm_Importance_sum=('Perm_Importance','sum'),
                       Perm_Std_sum=('Perm_Std','sum'),
                       Columns=('Column','count'))
                  .sort_values('Perm_Importance_sum', ascending=False))

print("MDI per column (top 20):\n", mdi.head(20))
print("\nPermutation per column (top 20):\n", perm_df.head(20))
print("\nMDI grouped:\n", mdi_grouped)
print("\nPermutation grouped:\n", perm_grouped)
```

```python
import pandas as pd
import numpy as np
import sklearn
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.feature_selection import RFE

# --- Load & normalize column names ---
df = pd.read_csv('/content/Final Primary_dataset_from_Indonesian_Private_University.csv')
df.columns = df.columns.str.strip().str.replace(r'\s+', ' ', regex=True)

# --- Your exact features & target ---
features = [
    'age', 'Gender', 'Work tenure', 'Education', 'Job position', 'working hour',
    'satisfaction with workload', 'satisfied compensation',
    'good relationship with peers', 'satisfied with career and opportunity',
    'satisfied with job profession', 'monthly salary',
    'satisfied with work-life balance', 'work is meaningful',
    'family support', 'mentally well and do not have anxiety'
]
target = 'TOI (turnover intention)'

# keep only these columns
df = df[features + [target]].dropna(subset=[target])
y = df[target].astype(int)

# split your feature types
categorical = ['age','Gender','Work tenure','Education','Job position','working hour','monthly salary']
numeric = [c for c in features if c not in categorical]

# ensure Likert numerics are numeric
for c in numeric:
    df[c] = pd.to_numeric(df[c], errors='coerce')
df = df.dropna(subset=features)

X = df[features]
y = df[target].astype(int)

# --- Version-safe OneHotEncoder (dense output) ---
ver = tuple(int(v) for v in sklearn.__version__.split('.')[:2])
ohe_kwargs = dict(drop='first', handle_unknown='ignore')
if ver >= (1, 2):
    ohe_kwargs['sparse_output'] = False
else:
    ohe_kwargs['sparse'] = False

ohe = OneHotEncoder(**ohe_kwargs)
scaler = StandardScaler()

preprocess = ColumnTransformer(
    transformers=[
        ('cat', ohe, categorical),
        ('num', scaler, numeric),
    ],
    remainder='drop'
)

# --- Base estimator for RFE ---
# (L2 is stable for ranking; class_weight helps imbalance)
est = LogisticRegression(
    solver='liblinear',
    penalty='l2',
    class_weight='balanced',
    max_iter=2000,
    random_state=42
)

# --- RFE: select top k; but we’ll inspect ranks for ALL columns ---
rfe = RFE(estimator=est, n_features_to_select=8, step=1)

pipe = Pipeline([('prep', preprocess), ('rfe', rfe)])
pipe.fit(X, y)

# --- Recover encoded feature names (dummies + numeric) ---
# refit an encoder on X[categorical] to get categories in consistent order
# Use the same kwargs we built earlier (already has drop + handle_unknown)
ohe_tmp = OneHotEncoder(**ohe_kwargs).fit(X[categorical])

cat_names = []
for feat, cats in zip(categorical, ohe_tmp.categories_):
    for cat in cats[1:]:  # drop='first' -> first category omitted
        cat_names.append(f"{feat}={cat}")


enc_feature_names = cat_names + numeric

# --- RFE outputs ---
support = pipe.named_steps['rfe'].support_          # True for selected
ranking = pipe.named_steps['rfe'].ranking_          # 1 = best (kept), 2,3,... eliminated earlier

rfe_results = pd.DataFrame({
    "Encoded_Feature": enc_feature_names,
    "Selected": support,
    "RFE_Rank": ranking
}).sort_values(["RFE_Rank","Encoded_Feature"]).reset_index(drop=True)

# --- Group back to original features: take best (lowest) rank among its columns ---
rfe_results['Original_Feature'] = rfe_results['Encoded_Feature'].str.split('=').str[0]
grouped = (rfe_results
           .groupby('Original_Feature', as_index=False)
           .agg(Best_Rank=('RFE_Rank','min'),
                Any_Selected=('Selected','max'),
                Columns=('Encoded_Feature','count'))
           .sort_values(['Best_Rank','Original_Feature'])
           .reset_index(drop=True))

print("\n=== RFE ranking (all encoded columns) ===")
print(rfe_results)

print("\n=== RFE grouped by original feature (best rank) ===")
print(grouped)
```

```python
# === ONE CELL to compute MF, ME, MW and merge ===
import numpy as np
import pandas as pd

from sklearn.feature_selection import chi2, f_classif, mutual_info_classif, RFE
from sklearn.linear_model import LogisticRegression, LogisticRegressionCV
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.pipeline import Pipeline
from sklearn.model_selection import StratifiedKFold
from sklearn.ensemble import RandomForestClassifier

# ------- Split features / target -------
X = df.iloc[:, 0:16].copy()
y = df.iloc[:, -1].copy()

# ------------------ MF (Chi2, ANOVA F, MI/IG) ------------------
chi_scores, _ = chi2(X, y)
f_scores, _   = f_classif(X, y)
mi_scores     = mutual_info_classif(X, y, discrete_features=True, random_state=42)

mf_tbl = pd.DataFrame({
    "Feature": X.columns,
    "Chi2": chi_scores,
    "ANOVA_F": f_scores,
    "Mutual_Info": mi_scores,
    "Info_Gain": mi_scores,
})

# Min–max normalize each metric
for c in ["Chi2","ANOVA_F","Mutual_Info","Info_Gain"]:
    cmin, cmax = mf_tbl[c].min(), mf_tbl[c].max()
    mf_tbl[c + "_Norm"] = 0.0 if cmax == cmin else (mf_tbl[c] - cmin) / (cmax - cmin)

mf_tbl["MF"] = mf_tbl[["Chi2_Norm","ANOVA_F_Norm","Mutual_Info_Norm","Info_Gain_Norm"]].mean(axis=1)
final_results_MF = mf_tbl[["Feature","MF"]]

# ------------------ ME (Lasso + RandomForest) ------------------
lasso_clf = Pipeline([
    ("scaler", StandardScaler()),
    ("logit", LogisticRegressionCV(
        Cs=np.logspace(-3, 3, 15),
        cv=StratifiedKFold(n_splits=5, shuffle=True, random_state=42),
        penalty="l1", solver="liblinear",
        scoring="f1_macro", max_iter=2000, n_jobs=-1, refit=True
    ))
])
lasso_clf.fit(X, y)

# Handle multi-class coefficients
coefs = np.abs(lasso_clf.named_steps["logit"].coef_)
if coefs.shape[0] > 1:  # multi-class
    lasso_coef = coefs.max(axis=0)
else:
    lasso_coef = coefs.ravel()

lasso_imp = pd.DataFrame({"Feature": X.columns, "Lasso_Importance": lasso_coef})

rf = RandomForestClassifier(n_estimators=500, class_weight="balanced", random_state=42, n_jobs=-1)
rf.fit(X, y)
rf_imp = pd.DataFrame({"Feature": X.columns, "RF_Importance": rf.feature_importances_})

me_tbl = lasso_imp.merge(rf_imp, on="Feature", how="inner")
for c in ["Lasso_Importance","RF_Importance"]:
    cmin, cmax = me_tbl[c].min(), me_tbl[c].max()
    me_tbl[c + "_Norm"] = 0.0 if cmax == cmin else (me_tbl[c] - cmin) / (cmax - cmin)
me_tbl["ME"] = me_tbl[["Lasso_Importance_Norm","RF_Importance_Norm"]].mean(axis=1)
final_results_ME = me_tbl[["Feature","ME"]]

# ------------------ MW (RFE + Boruta) ------------------
# RFE
base_est = LogisticRegression(solver="liblinear", class_weight="balanced", max_iter=2000, random_state=42)
rfe = RFE(estimator=base_est, n_features_to_select=8)
rfe.fit(X, y)
rfe_rank  = rfe.ranking_
rfe_score = rfe_rank.max() - rfe_rank + 1
rfe_tbl = pd.DataFrame({"Feature": X.columns, "RFE_Score": rfe_score})

# Boruta
boruta_ok = True
try:
    from boruta import BorutaPy
except Exception:
    try:
        from boruta_py import BorutaPy
    except Exception:
        boruta_ok = False

if boruta_ok:
    # Convert categorical columns to numeric if needed
    X_boruta = X.copy()
    for col in X_boruta.select_dtypes(include="category").columns:
        X_boruta[col] = LabelEncoder().fit_transform(X_boruta[col])

    y_boruta = y.copy()
    if y_boruta.dtype.name == "category" or y_boruta.dtype == object:
        y_boruta = LabelEncoder().fit_transform(y_boruta)

    rf_b = RandomForestClassifier(n_estimators=1000, class_weight="balanced",
                                  random_state=42, n_jobs=-1)
    boruta = BorutaPy(estimator=rf_b, n_estimators='auto', max_iter=200, random_state=42, verbose=0)
    boruta.fit(X_boruta.values, y_boruta)
    boruta_rank  = boruta.ranking_
    boruta_score = boruta_rank.max() - boruta_rank + 1
    boruta_tbl = pd.DataFrame({"Feature": X_boruta.columns, "Boruta_Score": boruta_score})
else:
    boruta_tbl = pd.DataFrame({"Feature": X.columns, "Boruta_Score": np.ones(X.shape[1])})

mw_tbl = rfe_tbl.merge(boruta_tbl, on="Feature", how="inner")
for c in ["RFE_Score","Boruta_Score"]:
    cmin, cmax = mw_tbl[c].min(), mw_tbl[c].max()
    mw_tbl[c + "_Norm"] = 0.0 if cmax == cmin else (mw_tbl[c] - cmin) / (cmax - cmin)
mw_tbl["MW"] = mw_tbl[["RFE_Score_Norm","Boruta_Score_Norm"]].mean(axis=1)
final_results_MW = mw_tbl[["Feature","MW"]]

# ------------------ Merge MF, ME, MW ------------------
all_scores = (
    final_results_MF
    .merge(final_results_ME, on="Feature", how="outer")
    .merge(final_results_MW, on="Feature", how="outer")
)

# Overall average & ranks
all_scores["Overall_Avg"]   = all_scores[["MF","ME","MW"]].mean(axis=1)
all_scores["Rank_Overall"]  = all_scores["Overall_Avg"].rank(ascending=False, method="dense").astype(int)

# Show
print(all_scores.sort_values("Overall_Avg", ascending=False).reset_index(drop=True))
```

## ENTRY FLOW AND LEAVING FLOW SUM FEATURE WISE AND PHI Calculation (feature wise)

```python
import pandas as pd
import itertools

# Step 0: Prepare MG/ME/MW from all_scores
df = all_scores.rename(columns={
    "MF": "MG",
    "ME": "ME",
    "MW": "MW"
})[["Feature","MG","ME","MW"]].reset_index(drop=True)

# Step 1: Build comparison table to find eliminated features
pairs = list(itertools.product(df["Feature"], df["Feature"]))
results = []

for a, b in pairs:
    a_vals = df.loc[df["Feature"] == a, ["MG", "ME", "MW"]].values[0]
    b_vals = df.loc[df["Feature"] == b, ["MG", "ME", "MW"]].values[0]

    mg_check = "✓" if a_vals[0] >= b_vals[0] else "✗"
    me_check = "✓" if a_vals[1] >= b_vals[1] else "✗"
    mw_check = "✓" if a_vals[2] >= b_vals[2] else "✗"

    wins = sum([mg_check=="✓", me_check=="✓", mw_check=="✓"])
    c_ab = wins / 3
    decision = "Yes" if c_ab >= 0.6599 else "No"

    results.append([a, b, mg_check, me_check, mw_check, wins, round(c_ab, 4), decision])

comparison_table = pd.DataFrame(
    results, columns=["A", "B", "MG", "ME", "MW", "Wins", "C(a,b)", "Decision"]
)

# Step 2: Get only remaining features
eliminate_features = [
    feature for feature in df["Feature"]
    if all(
        comparison_table[(comparison_table["A"] == feature) &
                         (comparison_table["A"] != comparison_table["B"])]["Decision"] == "No"
    )
]

filtered_df = df[~df["Feature"].isin(eliminate_features)].reset_index(drop=True)

# Step 3: Compute pi(a,b) for all pairs (a != b) with debug
pi_results = []

for entry in filtered_df["Feature"]:
    entry_vals = filtered_df.loc[filtered_df["Feature"] == entry, ["MG", "ME", "MW"]].values[0]

    for leaving in filtered_df["Feature"]:
        if entry == leaving:
            continue

        leaving_vals = filtered_df.loc[filtered_df["Feature"] == leaving, ["MG", "ME", "MW"]].values[0]

        # Direct differences (can be negative)
        mg_diff = entry_vals[0] - leaving_vals[0]
        me_diff = entry_vals[1] - leaving_vals[1]
        mw_diff = entry_vals[2] - leaving_vals[2]

        pi_val = round((mg_diff + me_diff + mw_diff) / 3, 4)

        # Debug string
        calc_str = f"(({entry_vals[0]}-{leaving_vals[0]}) + ({entry_vals[1]}-{leaving_vals[1]}) + ({entry_vals[2]}-{leaving_vals[2]})) / 3 = {pi_val}"

        pi_results.append({
            "Entry Flow": entry,
            "Leaving Flow": leaving,
            "MG_diff": round(mg_diff, 4),
            "ME_diff": round(me_diff, 4),
            "MW_diff": round(mw_diff, 4),
            "pi": pi_val,
            "Calculation": calc_str
        })

pi_df = pd.DataFrame(pi_results)

# Step 4: Compute phi+ and phi- for each feature
phi_plus = pi_df.groupby("Entry Flow")["pi"].mean().rename("phi+")
phi_minus = pi_df.groupby("Leaving Flow")["pi"].mean().rename("phi-")

phi_table = pd.concat([phi_plus, phi_minus], axis=1).reset_index().rename(columns={"index":"Feature"})

# Step 5: Compute Net Flow (phi = phi+ - phi-)
phi_table["phi"] = (phi_table["phi+"] - phi_table["phi-"]).round(4)

# Step 6: Rank features by Net Flow (descending)
phi_table["Rank"] = phi_table["phi"].rank(method="dense", ascending=False).astype(int)

print("\nPairwise Preference Index (pi with full calculation):\n")
print(pi_df.to_string(index=False))

print("\nPhi+ , Phi- , Net Flow (phi) and Ranking:\n")
print(phi_table.sort_values("Rank").to_string(index=False))
```
