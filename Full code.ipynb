```python
import pandas as pd
import numpy as np

# --- Load & normalize column names ---
file_path = '/content/Final Primary_dataset_from_Indonesian_Private_University.csv'  # adjust if needed
df = pd.read_csv(file_path)
df.columns = df.columns.str.strip().str.replace(r'\s+', ' ', regex=True)

# --- Columns of interest ---
target = 'TOI (turnover intention)'
cols = [
    'satisfaction with workload',
    'good relationship with peers',
    'family support'
]

# --- Keep needed columns, coerce to numeric, drop missing ---
df_small = df[cols + [target]].copy()
for c in cols:
    df_small[c] = pd.to_numeric(df_small[c], errors='coerce')
df_small[target] = pd.to_numeric(df_small[target], errors='coerce').astype('Int64')
df_small = df_small.dropna(subset=cols + [target])
df_small[target] = df_small[target].astype(int)

def toi_freq_table(col_name: str) -> pd.DataFrame:
    # counts of TOI=0/1 by each Likert value
    ct = pd.crosstab(df_small[col_name], df_small[target]).rename(columns={0: 'TOI=0', 1: 'TOI=1'})
    if 'TOI=0' not in ct.columns: ct['TOI=0'] = 0
    if 'TOI=1' not in ct.columns: ct['TOI=1'] = 0
    ct = ct[['TOI=0', 'TOI=1']].sort_index()
    ct['Total'] = ct['TOI=0'] + ct['TOI=1']
    ct['TOI=1 rate (%)'] = (ct['TOI=1'] / ct['Total'] * 100).round(2)
    ct.index.name = f'{col_name} (value)'
    return ct.reset_index()

# --- Compute & print tables ---
tables = {c: toi_freq_table(c) for c in cols}
for name, table in tables.items():
    print(f'\n=== TOI frequency by value — {name} ===')
    print(table.to_string(index=False))

# --- (Optional) Save to CSVs ---
# for name, table in tables.items():
#     safe = name.replace(' ', '_')
#     table.to_csv(f'{safe}_TOI_frequency.csv', index=False)
```

```python
import pandas as pd
import numpy as np
import scipy.stats as stats

# --- Load dataset ---
file_path = '/content/Final Primary_dataset_from_Indonesian_Private_University.csv'
df = pd.read_csv(file_path)

# --- Clean column names (strip + collapse spaces) ---
df.columns = df.columns.str.strip().str.replace(r'\s+', ' ', regex=True)

# --- Features & target ---
features = [
    'age', 'Gender', 'Work tenure', 'Education', 'Job position', 'working hour',
    'satisfaction with workload', 'satisfied compensation',
    'good relationship with peers', 'satisfied with career and opportunity',
    'satisfied with job profession', 'monthly salary',
    'satisfied with work-life balance', 'work is meaningful',
    'family support', 'mentally well and do not have anxiety'
]
target = 'TOI (turnover intention)'

# --- Simple binning for continuous variables ---
continuous = ['age', 'Work tenure', 'working hour', 'monthly salary']
for col in continuous:
    if col in df.columns and pd.api.types.is_numeric_dtype(df[col]):
        # 4 bins (quartiles); adjust if you want specific ranges
        df[col+'_bin'] = pd.qcut(df[col], q=4, duplicates='drop')
        features[features.index(col)] = col+'_bin'   # replace with binned version

# --- Chi-Square calculation ---
results = []
for col in features:
    if col not in df.columns:
        continue
    contingency = pd.crosstab(df[col], df[target])
    chi2, p, _, _ = stats.chi2_contingency(contingency)
    results.append([col, chi2, p])

# --- Results table ---
chi_square_table = pd.DataFrame(results, columns=["Feature", "Chi2", "p-value"])\
                     .sort_values(by="Chi2", ascending=False)\
                     .reset_index(drop=True)

print(chi_square_table)
```

```python
# --- Setup ---
import pandas as pd
import numpy as np
from sklearn.feature_selection import f_classif

# 1) Load your CSV (adjust path if needed)
file_path = '/content/Final Primary_dataset_from_Indonesian_Private_University.csv'
df = pd.read_csv(file_path)

# 2) Clean column names: strip, collapse multiple spaces
df.columns = (
    df.columns
      .str.strip()
      .str.replace(r'\s+', ' ', regex=True)
)

# 3) Define features + target
features = [
    'age', 'Gender', 'Work tenure', 'Education', 'Job position', 'working hour',
    'satisfaction with workload', 'satisfied compensation',
    'good relationship with peers', 'satisfied with career and opportunity',
    'satisfied with job profession', 'monthly salary',
    'satisfied with work-life balance', 'work is meaningful',
    'family support', 'mentally well and do not have anxiety'
]
target = 'TOI (turnover intention)'

# 4) Keep only your features + target
df = df[features + [target]].copy()

# 5) Drop rows with missing target & convert to numeric
df = df.dropna(subset=[target])
df[target] = pd.to_numeric(df[target], errors='coerce').astype(int)

# 6) Split X and y
X_raw = df[features].copy()
y = df[target]

# 7) Identify numeric vs categorical
numeric_cols = X_raw.select_dtypes(include=[np.number]).columns.tolist()
categorical_cols = [c for c in X_raw.columns if c not in numeric_cols]

# 8) One-hot encode categoricals (safe separator)
SAFE_SEP = '§'
X_cat = pd.get_dummies(
    X_raw[categorical_cols],
    prefix=categorical_cols,
    prefix_sep=SAFE_SEP,
    drop_first=True,   # drop baseline level
    dtype=np.uint8
) if categorical_cols else pd.DataFrame(index=X_raw.index)

# 9) Combine numeric + encoded categoricals
X_num = X_raw[numeric_cols].copy()
X = pd.concat([X_num, X_cat], axis=1)

# Drop any rows with NaNs in features
mask = X.notna().all(axis=1)
X = X.loc[mask]
y = y.loc[mask]

# 10) Run ANOVA F-test
f_values, p_values = f_classif(X.values, y.values)

per_column = pd.DataFrame({
    'Column': X.columns,
    'F_value': f_values,
    'p_value': p_values
}).sort_values('F_value', ascending=False).reset_index(drop=True)

# 11) Aggregate back to original features
def original_feature(col: str) -> str:
    return col.split(SAFE_SEP, 1)[0] if SAFE_SEP in col else col

per_column['Feature'] = per_column['Column'].map(original_feature)

agg = (per_column
       .groupby('Feature', as_index=False)
       .agg(F_value_max=('F_value', 'max'),
            p_value_min=('p_value', 'min'))
       .sort_values('F_value_max', ascending=False)
       .reset_index(drop=True))


print("\n ANOVA F ")
print(agg)
```

```python
# ===== LASSO (L1 Logistic) feature selection for TOI =====
# - Uses ONLY your declared features + target
# - One-hot encodes categoricals, scales Likert numerics
# - Cross-validated strength of regularization (C)
# - Produces per-column and grouped-by-feature tables and saves as CSVs

import pandas as pd
import numpy as np
import sklearn
from sklearn.model_selection import StratifiedKFold
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegressionCV

print("scikit-learn version:", sklearn.__version__)

# ---------- 1) Load & normalize column names ----------
file_path = '/content/Final Primary_dataset_from_Indonesian_Private_University.csv'  # <-- adjust if needed
df = pd.read_csv(file_path)
# strip ends & collapse multiple spaces to a single space (your CSV has extra spaces)
df.columns = df.columns.str.strip().str.replace(r'\s+', ' ', regex=True)

# ---------- 2) Your features and target (normalized names) ----------
features = [
    'age', 'Gender', 'Work tenure', 'Education', 'Job position', 'working hour',
    'satisfaction with workload', 'satisfied compensation',
    'good relationship with peers', 'satisfied with career and opportunity',
    'satisfied with job profession', 'monthly salary',
    'satisfied with work-life balance', 'work is meaningful',
    'family support', 'mentally well and do not have anxiety'
]
target = 'TOI (turnover intention)'

# keep only your columns; drop rows with missing target
missing = [c for c in features + [target] if c not in df.columns]
if missing:
    raise KeyError(f"Missing columns after normalization: {missing}")

df = df[features + [target]].dropna(subset=[target])
y = df[target].astype(int)

# ---------- 3) Split numeric (Likert) vs categorical ----------
categorical_features = [
    'age', 'Gender', 'Work tenure', 'Education', 'Job position',
    'working hour', 'monthly salary'
]
numeric_features = [c for c in features if c not in categorical_features]

# ensure Likert numerics are numeric
for col in numeric_features:
    df[col] = pd.to_numeric(df[col], errors='coerce')

# drop rows with any missing in selected features
df = df.dropna(subset=features)
X = df[features]
y = df[target].astype(int)

# ---------- 4) Preprocessing: OHE for categoricals + scaling for numerics ----------
SAFE_SEP = '§'  # used to rebuild original feature names from dummies later

# Handle OneHotEncoder API differences across sklearn versions
# - <1.2: OneHotEncoder(..., sparse=False)
# - >=1.2: OneHotEncoder(..., sparse_output=False)
ohe_kwargs = dict(drop='first', handle_unknown='ignore')
ver = tuple(int(v) for v in sklearn.__version__.split('.')[:2])
if ver >= (1, 2):
    ohe_kwargs['sparse_output'] = False
else:
    ohe_kwargs['sparse'] = False

ohe = OneHotEncoder(**ohe_kwargs)
scaler = StandardScaler()

preprocess = ColumnTransformer(
    transformers=[
        ('cat', ohe, categorical_features),
        ('num', scaler, numeric_features),
    ],
    remainder='drop'
)

# ---------- 5) L1-penalized Logistic Regression with CV ----------
# We optimize F1 and use class_weight='balanced' for any class imbalance.
logreg_cv = LogisticRegressionCV(
    Cs=np.logspace(-2, 1, 8),  # search range for C (inverse of regularization)
    cv=StratifiedKFold(n_splits=3, shuffle=True, random_state=42),
    penalty='l1',
    solver='saga',             # supports l1 with class_weight
    scoring='f1',
    class_weight='balanced',
    max_iter=4000,
    n_jobs=-1,
    refit=True
)

pipe = Pipeline(steps=[('prep', preprocess), ('clf', logreg_cv)])
pipe.fit(X, y)

# ---------- 6) Recover feature names after preprocessing ----------
# Build Feature§Category names for one-hot columns so we can group later.
cat_feature_names = []
ohe_fitted = pipe.named_steps['prep'].named_transformers_['cat']
for feat, cats in zip(categorical_features, ohe_fitted.categories_):
    # drop='first' => first category is the baseline and not present in output
    for cat in cats[1:]:
        cat_feature_names.append(f"{feat}{SAFE_SEP}{cat}")

num_feature_names = numeric_features[:]  # unchanged by StandardScaler
all_feature_names = cat_feature_names + num_feature_names

# ---------- 7) Coefficients (LASSO shrinkage; zeros mean dropped) ----------
coefs = pipe.named_steps['clf'].coef_.ravel()
per_column = pd.DataFrame({
    'Column': all_feature_names,
    'Coefficient': coefs
})
per_column['Abs_Coeff'] = per_column['Coefficient'].abs()
per_column = per_column.sort_values('Abs_Coeff', ascending=False).reset_index(drop=True)

def original_feature_name(col: str) -> str:
    return col.split(SAFE_SEP, 1)[0] if SAFE_SEP in col else col

per_column['Feature'] = per_column['Column'].map(original_feature_name)

# Aggregate back to original feature:
# - Importance = sum of |coefficients| over its columns
# - NonZero_Columns = count of nonzero coefficients (survived lasso)
# - Max_Abs = largest |coef| among its columns
# - Columns = total columns contributed (1 for numeric; >1 for categoricals)
grouped = (
    per_column
    .groupby('Feature', as_index=False)
    .agg(Importance=('Abs_Coeff', 'sum'),
         NonZero_Columns=('Coefficient', lambda x: int(np.sum(x != 0))),
         Max_Abs=('Abs_Coeff', 'max'),
         Columns=('Column', 'count'))
    .sort_values('Importance', ascending=False)
    .reset_index(drop=True)
)

# ---------- 8) Show & save ----------
print("\n=== L1-Logistic (LASSO) — Per Column Coefficients (top 25) ===")
print(per_column.head(25))
print("\n=== L1-Logistic (LASSO) — Grouped by Feature ===")
print(grouped)


# Optional: quick glance at class balance
print("\nClass counts (TOI):")
print(y.value_counts())
```

```python
import pandas as pd
import re
import numpy as np
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.linear_model import LogisticRegressionCV
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer

# --- Load dataset ---
df = pd.read_csv('/content/Final Primary_dataset_from_Indonesian_Private_University.csv')
df.columns = df.columns.str.strip().str.replace(r'\s+', ' ', regex=True)

# --- Features & target ---
features = [
    'age', 'Gender', 'Work tenure', 'Education', 'Job position', 'working hour',
    'satisfaction with workload', 'satisfied compensation',
    'good relationship with peers', 'satisfied with career and opportunity',
    'satisfied with job profession', 'monthly salary',
    'satisfied with work-life balance', 'work is meaningful',
    'family support', 'mentally well and do not have anxiety'
]
target = 'TOI (turnover intention)'

df = df[features + [target]].dropna(subset=[target])
y = df[target].astype(int)

# --- Clean numeric features ---
def extract_number(s):
    if pd.isna(s): return np.nan
    s = str(s).replace(',', '')
    m = re.search(r'(-?\d+(?:\.\d+)?)', s)
    return float(m.group(1)) if m else np.nan

for col in ['age','Work tenure','working hour','monthly salary']:
    df[col] = df[col].apply(extract_number)

numeric_features = ['age','Work tenure','working hour','monthly salary']
categorical_features = [c for c in features if c not in numeric_features]

# --- Preprocessing ---
numeric_tf = Pipeline([
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])
categorical_tf = Pipeline([
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('onehot', OneHotEncoder(handle_unknown='ignore', drop='first'))
])
preprocessor = ColumnTransformer([
    ('num', numeric_tf, numeric_features),
    ('cat', categorical_tf, categorical_features)
])

# --- Logistic Regression with L1 penalty (LASSO) ---
lasso = LogisticRegressionCV(
    Cs=10,
    penalty="l1",
    solver="liblinear",
    cv=5,
    class_weight="balanced",
    max_iter=2000,
    random_state=42
)

pipe = Pipeline([
    ('prep', preprocessor),
    ('clf', lasso)
])

# --- Fit ---
pipe.fit(df[features], y)

# --- Extract encoded feature names & coefficients ---
feature_names = pipe.named_steps['prep'].get_feature_names_out()
coefs = pipe.named_steps['clf'].coef_.ravel()

lasso_results = pd.DataFrame({
    'Encoded_Feature': feature_names,
    'Coefficient': coefs
})

# --- Map encoded → original feature ---
def to_original(name):
    # e.g. "cat__Job position_Manager" → "Job position"
    if "__" in name:
        after = name.split("__", 1)[1]   # drop "num__" or "cat__"
        return after.split("_", 1)[0]    # take only the feature name before first underscore
    return name

lasso_results['Original_Feature'] = lasso_results['Encoded_Feature'].apply(to_original)

# --- Group back to 16 original features ---
grouped = (
    lasso_results.groupby('Original_Feature', as_index=False)
    .agg(
        # max abs coefficient across dummies = feature strength
        MaxAbsCoef=('Coefficient', lambda x: np.max(np.abs(x))),
        Any_Selected=('Coefficient', lambda x: (x != 0).any()),
        Num_Encoded=('Encoded_Feature','count')
    )
    .sort_values('MaxAbsCoef', ascending=False)
    .reset_index(drop=True)
)

# Add a normalized score for easy comparison
grouped['Score_Norm'] = grouped['MaxAbsCoef'] / grouped['MaxAbsCoef'].max()

print("\n=== LASSO Logistic Regression Grouped by Original Feature ===")
print(grouped[['Original_Feature','MaxAbsCoef','Score_Norm','Any_Selected','Num_Encoded']])
```

```python
import pandas as pd
import re
import numpy as np
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.model_selection import train_test_split

# --- Load dataset ---
df = pd.read_csv('/content/Final Primary_dataset_from_Indonesian_Private_University.csv')
df.columns = df.columns.str.strip().str.replace(r'\s+', ' ', regex=True)

# --- Features & target ---
features = [
    'age', 'Gender', 'Work tenure', 'Education', 'Job position', 'working hour',
    'satisfaction with workload', 'satisfied compensation',
    'good relationship with peers', 'satisfied with career and opportunity',
    'satisfied with job profession', 'monthly salary',
    'satisfied with work-life balance', 'work is meaningful',
    'family support', 'mentally well and do not have anxiety'
]
target = 'TOI (turnover intention)'

df = df[features + [target]].dropna(subset=[target])
y = df[target].astype(int)

# --- Clean numeric features ---
def extract_number(s):
    if pd.isna(s): return np.nan
    s = str(s).replace(',', '')
    m = re.search(r'(-?\d+(?:\.\d+)?)', s)
    return float(m.group(1)) if m else np.nan

for col in ['age','Work tenure','working hour','monthly salary']:
    df[col] = df[col].apply(extract_number)

numeric_features = ['age','Work tenure','working hour','monthly salary']
categorical_features = [c for c in features if c not in numeric_features]

# --- Preprocessing ---
numeric_tf = Pipeline([
    ('imputer', SimpleImputer(strategy='median'))
])
categorical_tf = Pipeline([
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('onehot', OneHotEncoder(handle_unknown='ignore'))
])
preprocessor = ColumnTransformer([
    ('num', numeric_tf, numeric_features),
    ('cat', categorical_tf, categorical_features)
])

# --- Random Forest pipeline ---
rf = Pipeline([
    ('prep', preprocessor),
    ('clf', RandomForestClassifier(n_estimators=200, random_state=42, class_weight='balanced'))
])

# --- Train/Test Split ---
X_train, X_test, y_train, y_test = train_test_split(df[features], y, test_size=0.3, random_state=42)

# --- Fit ---
rf.fit(X_train, y_train)

# --- Extract feature importances (dummy level) ---
model = rf.named_steps['clf']
encoded_names = rf.named_steps['prep'].get_feature_names_out()
importances = model.feature_importances_

importance_df = pd.DataFrame({
    'Encoded_Feature': encoded_names,
    'Importance': importances
})

# --- Map encoded → original feature ---
def to_original(name):
    # e.g. "cat__Job position_Manager" → "Job position"
    if "__" in name:
        after = name.split("__", 1)[1]
        return after.split("_", 1)[0]  # take only the feature name
    return name

importance_df['Original_Feature'] = importance_df['Encoded_Feature'].apply(to_original)

# --- Group back to original 16 features ---
grouped = (
    importance_df.groupby('Original_Feature', as_index=False)
    .agg(
        RF_Importance_Sum=('Importance','sum'),
        Num_Encoded=('Encoded_Feature','count')
    )
    .sort_values('RF_Importance_Sum', ascending=False)
    .reset_index(drop=True)
)

# Normalize importance
grouped['RF_Importance_Norm'] = grouped['RF_Importance_Sum'] / grouped['RF_Importance_Sum'].max()
grouped['Feature_Rank'] = range(1, len(grouped)+1)

print("\n=== Random Forest Feature Importance (Grouped by Original Feature) ===")
print(grouped[['Feature_Rank','Original_Feature','RF_Importance_Sum','RF_Importance_Norm','Num_Encoded']])
```

```python
import pandas as pd
import numpy as np
import re
from sklearn.feature_selection import RFE
from sklearn.linear_model import LogisticRegression
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline

# --- Load dataset ---
df = pd.read_csv('/content/Final Primary_dataset_from_Indonesian_Private_University.csv')
df.columns = df.columns.str.strip().str.replace(r'\s+', ' ', regex=True)

# --- Define features & target ---
features = [
    'age', 'Gender', 'Work tenure', 'Education', 'Job position', 'working hour',
    'satisfaction with workload', 'satisfied compensation',
    'good relationship with peers', 'satisfied with career and opportunity',
    'satisfied with job profession', 'monthly salary',
    'satisfied with work-life balance', 'work is meaningful',
    'family support', 'mentally well and do not have anxiety'
]
target = 'TOI (turnover intention)'

df = df[features + [target]].dropna(subset=[target])
y = df[target].astype(int)

# --- Clean numeric columns ---
def extract_number(s):
    """Extract the first number from a string, else NaN."""
    if pd.isna(s):
        return np.nan
    s = str(s).replace(',', '')
    m = re.search(r'(-?\d+(?:\.\d+)?)', s)
    return float(m.group(1)) if m else np.nan

for col in ['age', 'Work tenure', 'working hour', 'monthly salary']:
    if col in df.columns:
        df[col] = df[col].apply(extract_number)

# --- Feature types ---
numeric_features = ['age', 'Work tenure', 'working hour', 'monthly salary']
categorical_features = [c for c in features if c not in numeric_features]

# --- Preprocessing ---
num_tf = Pipeline([
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])
cat_tf = Pipeline([
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('onehot', OneHotEncoder(handle_unknown='ignore', drop='first'))
])

preprocessor = ColumnTransformer([
    ('num', num_tf, numeric_features),
    ('cat', cat_tf, categorical_features),
], remainder='drop', verbose_feature_names_out=True)

# --- Logistic Regression model ---
estimator = LogisticRegression(
    solver='liblinear',
    max_iter=2000,
    class_weight='balanced',
    random_state=42
)

# --- Wrap preprocessing + RFE ---
pipe = Pipeline([
    ('prep', preprocessor),
    ('rfe', RFE(estimator=estimator, n_features_to_select=8))
])

# --- Fit ---
pipe.fit(df[features], y)

# --- Get encoded feature names ---
encoded_names = pipe.named_steps['prep'].get_feature_names_out()

# --- Collect RFE results ---
rfe = pipe.named_steps['rfe']
rfe_results = pd.DataFrame({
    'Encoded_Feature': encoded_names,
    'Selected': rfe.support_,
    'Rank': rfe.ranking_
})

# --- Map back to original features ---
def to_original(name):
    after = name.split('__', 1)[1]   # drop "num__" or "cat__"
    return after.split('_', 1)[0]    # e.g. "Job position_Manager" → "Job position"

rfe_results['Original_Feature'] = rfe_results['Encoded_Feature'].apply(to_original)

# --- Aggregate per original feature ---
grouped = (
    rfe_results
    .groupby('Original_Feature', as_index=False)
    .agg(
        Best_Rank=('Rank', 'min'),
        Any_Selected=('Selected', 'max'),
        Columns=('Encoded_Feature', 'count')
    )
    .sort_values('Best_Rank', kind='mergesort')
    .reset_index(drop=True)
)

# Re-index 1..16 for clean ranking
grouped['Feature_Rank'] = range(1, len(grouped) + 1)

print("\n=== RFE Grouped by Original Feature ===")
print(grouped[['Feature_Rank','Original_Feature','Best_Rank','Any_Selected','Columns']])
```

```python
import pandas as pd
import numpy as np
import re
from sklearn.ensemble import RandomForestClassifier
from boruta import BorutaPy
from sklearn.preprocessing import OneHotEncoder

# --- Load dataset ---
df = pd.read_csv('/content/Final Primary_dataset_from_Indonesian_Private_University.csv')
df.columns = df.columns.str.strip().str.replace(r'\s+', ' ', regex=True)

# --- Features & target ---
features = [
    'age', 'Gender', 'Work tenure', 'Education', 'Job position', 'working hour',
    'satisfaction with workload', 'satisfied compensation',
    'good relationship with peers', 'satisfied with career and opportunity',
    'satisfied with job profession', 'monthly salary',
    'satisfied with work-life balance', 'work is meaningful',
    'family support', 'mentally well and do not have anxiety'
]
target = 'TOI (turnover intention)'

df = df[features + [target]].dropna(subset=[target])
y = df[target].astype(int)

# --- Clean numeric columns ---
def extract_number(s):
    if pd.isna(s): return np.nan
    s = str(s).replace(',', '')
    m = re.search(r'(-?\d+(?:\.\d+)?)', s)
    return float(m.group(1)) if m else np.nan

for col in ['age','Work tenure','working hour','monthly salary']:
    df[col] = df[col].apply(extract_number)

# --- Split categorical vs numeric ---
cat_features = [c for c in features if c not in ['age','Work tenure','working hour','monthly salary']]
num_features = ['age','Work tenure','working hour','monthly salary']

# --- One-hot encode categorical ---
ohe = OneHotEncoder(handle_unknown='ignore', drop='first', sparse_output=False)
X_cat = ohe.fit_transform(df[cat_features])
cat_names = ohe.get_feature_names_out(cat_features)

# Combine numeric + categorical
X_num = df[num_features].to_numpy()
X = np.hstack([X_num, X_cat])
feature_names = num_features + list(cat_names)

# --- RandomForest model for Boruta ---
rf = RandomForestClassifier(
    n_estimators=100,
    n_jobs=-1,
    class_weight='balanced',
    random_state=42
)

# --- Run Boruta ---
feat_selector = BorutaPy(
    estimator=rf,
    n_estimators='auto',
    max_iter=20,
    verbose=2,
    random_state=42
)
feat_selector.fit(X, y.values)

# --- Collect Boruta results (encoded) ---
boruta_df = pd.DataFrame({
    'Feature_Encoded': feature_names,
    'Boruta_Support': feat_selector.support_,
    'Boruta_Ranking': feat_selector.ranking_
})

# --- Fit RF separately for importances ---
rf.fit(X, y)
importances = rf.feature_importances_

importance_df = pd.DataFrame({
    'Feature_Encoded': feature_names,
    'RF_Importance': importances
})

# --- Merge Boruta + RF ---
merged = boruta_df.merge(importance_df, on='Feature_Encoded', how='left')

# --- Map encoded → original feature ---
def to_original(name):
    # e.g. "Job position_Manager" → "Job position"
    return name.split('_', 1)[0] if '_' in name else name

merged['Original_Feature'] = merged['Feature_Encoded'].apply(to_original)

# --- Group back to original 16 features ---
grouped = (
    merged.groupby('Original_Feature', as_index=False)
    .agg(
        Boruta_Group_Support=('Boruta_Support', 'max'),
        Boruta_Best_Rank=('Boruta_Ranking', 'min'),
        RF_Importance_Sum=('RF_Importance', 'sum')
    )
)

# Normalize RF importance
grouped['RF_Importance_Norm'] = grouped['RF_Importance_Sum'] / grouped['RF_Importance_Sum'].max()

# Sort by Boruta rank, then importance
grouped = grouped.sort_values(['Boruta_Best_Rank','RF_Importance_Sum'],
                              ascending=[True,False]).reset_index(drop=True)

# Re-rank 1..16
grouped['Feature_Rank'] = range(1, len(grouped)+1)

print("\n=== Grouped Boruta + RF Feature Importance (16 features) ===")
print(grouped[['Feature_Rank','Original_Feature',
               'Boruta_Group_Support','Boruta_Best_Rank',
               'RF_Importance_Sum','RF_Importance_Norm']])
```
